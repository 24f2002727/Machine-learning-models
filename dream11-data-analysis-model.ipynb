```python
import pandas as pd
```

```python
bat_data=pd.read_csv("/content/Batting_data.csv")
bat_data.head()
```

```python
field_data=pd.read_csv("/content/Fielding_data.csv")
field_data.head()
```

```python
final_data=pd.read_csv("/content/Final_Fantasy_data.csv")
final_data.head()
```

```python
bow_data=pd.read_csv("/content/Bowling_data.csv")
bow_data.head()
```

```python
match_data=pd.read_csv("/content/Match_details.csv")
match_data.head()
```

```python
data=pd.read_csv("/content/data.csv")
data.head()
```

```python
#Top batters code
top_batters=(bat_data.groupby('fullName')['runs'].sum().sort_values(ascending=False).reset_index().head(10))
print(top_batters)
```

```python
#Top bowlers
top_bowlers=(bow_data.groupby('fullName')['wickets'].sum().sort_values(ascending=False).reset_index().head(10))
print(top_bowlers)
```

```python
#Top fielders
top_fielders=(field_data.groupby('fullName')['Fielding_FP'].sum().sort_values(ascending=False).reset_index().head(10))
print(top_fielders)
```

```python
#Top Fantasy player
fantaty_player=(final_data.groupby('fullName')['Total_FP'].sum().sort_values(ascending=False).reset_index().head(20))
print(fantaty_player)
```
